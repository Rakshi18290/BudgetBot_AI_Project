```python

# !pip install pandas numpy scikit-learn matplotlib seaborn joblib xgboost shap --quiet
import os, sys, math, json, joblib
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import xgboost as xgb
from xgboost import XGBRegressor

# paths
PROJECT_ROOT = os.path.expanduser('~/Documents/BudgetBot_AI_Project1')
DATA_DIR = r"C:\Users\admin\Desktop\BudgetBot_AI_Project1\BudgetBot_AI_Project1\BudgetBot_AI_Project1\data"
#DATA_DIR = os.path.join(PROJECT_ROOT, 'data')
MODELS_DIR = os.path.join(PROJECT_ROOT, 'models')
RESULTS_DIR = os.path.join(PROJECT_ROOT, 'results')
os.makedirs(DATA_DIR, exist_ok=True)
os.makedirs(MODELS_DIR, exist_ok=True)
os.makedirs(RESULTS_DIR, exist_ok=True)

print('Environment ready.')
```

## 2) Data Loading & Inspection
Load your CSV (place it at `/mnt/data/data (2).csv`). Inspect shape, columns, dtypes and head.

```python

csv_path = os.path.join(DATA_DIR, 'data (2).csv')
if not os.path.exists(csv_path):
    raise FileNotFoundError(f"Place your dataset at: {csv_path}")
df = pd.read_csv(csv_path)
print('Shape:', df.shape)
print('\nColumns:', df.columns.tolist())
print('\nDtypes:\n', df.dtypes)
print('\nMissing values:\n', df.isnull().sum())
df.head()
```

## 3) Preprocessing & Feature Engineering
Create derived features: total_expenses, debt_to_income_ratio, expense_ratio, savings_estimate, Loan_Repayment (EMI if missing). Save train/test parquet files.

```python

# ✅ Basic cleaning / standardize column names
df.columns = [c.strip() for c in df.columns]

# ✅ Handle missing debt column safely
if 'debt' not in df.columns:
    df['debt'] = 0.0  # default to 0 if not available

# ✅ Detect expense columns
expense_cols = [c for c in df.columns if c.lower() in (
    'rent','groceries','transport','utilities','healthcare','education',
    'eating_out','entertainment','insurance','miscellaneous','loan_repayment'
)]

if set(['Rent','Groceries','Transport','Utilities']).intersection(df.columns):
    df['total_expenses'] = df[['Rent','Groceries','Transport','Utilities']].sum(axis=1)
elif len(expense_cols) > 0:
    df['total_expenses'] = df[expense_cols].sum(axis=1)
else:
    if 'total_expenses' not in df.columns:
        df['total_expenses'] = 0.0

# ✅ Derived financial ratios
df['Income'] = pd.to_numeric(df['Income'], errors='coerce').fillna(0)
df['debt'] = pd.to_numeric(df['debt'], errors='coerce').fillna(0)

df['debt_to_income_ratio'] = (df['debt'] / df['Income'].replace(0, np.nan)).replace([np.inf, -np.inf], 0).fillna(0)
df['expense_ratio'] = (df['total_expenses'] / df['Income'].replace(0, np.nan)).replace([np.inf, -np.inf], 0).fillna(0)
df['savings_estimate'] = df['Income'] - df['total_expenses'] - df['debt']*0.12

# ✅ EMI estimator function
def compute_emi(principal, annual_rate=0.12, years=3):
    try:
        p = float(principal)
    except:
        return 0.0
    if p <= 0:
        return 0.0
    r = annual_rate / 12.0
    n = years * 12
    emi = (p*r*(1+r)**n) / ((1+r)**n - 1)
    return round(emi, 2)

# ✅ Add Loan_Repayment if not present
if 'Loan_Repayment' not in df.columns:
    df['Loan_Repayment'] = df['debt'].apply(lambda d: compute_emi(d))

# ✅ Convert expected numeric columns
numeric_cols = ['Age', 'Income', 'Dependents', 'debt', 
                'debt_to_income_ratio', 'expense_ratio', 
                'Loan_Repayment', 'total_expenses']

for c in numeric_cols:
    if c in df.columns:
        df[c] = pd.to_numeric(df[c], errors='coerce').fillna(0)

# ✅ Fill any remaining NaNs
df.fillna(0, inplace=True)

# ✅ Train/Test split + save
train, test = train_test_split(df, test_size=0.2, random_state=42)
train.to_parquet(os.path.join(DATA_DIR, 'train.parquet'), index=False)
test.to_parquet(os.path.join(DATA_DIR, 'test.parquet'), index=False)
df.to_parquet(os.path.join(DATA_DIR, 'preprocessed.parquet'), index=False)

print("✅ Preprocessing complete.")
print("Train shape:", train.shape)
print("Test shape:", test.shape)
print("Saved preprocessed files to:", DATA_DIR)
```

## 4) Preprocessor Pipeline
Define numeric and categorical features, build ColumnTransformer and save it.

```python

# Feature & target specification (adjust if your column names differ)
FEATURES = ['Age','Income','Dependents','debt','debt_to_income_ratio','expense_ratio','Loan_Repayment']
CATEGORICAL = ['Occupation','City_Tier']
# Ensure categorical exist
for c in CATEGORICAL:
    if c not in train.columns:
        print(f"Warning: {c} not in data columns; removing from categorical list.")
        CATEGORICAL.remove(c)

TARGETS = ['Rent','Groceries','Transport','Utilities','Healthcare','Education','Eating_Out','Entertainment','Loan_Repayment','Insurance','Miscellaneous','Savings']
# Keep only targets that exist in dataframe
TARGETS = [t for t in TARGETS if t in train.columns]

from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer

num_cols = FEATURES
cat_cols = CATEGORICAL

preprocessor = ColumnTransformer(transformers=[
    ('num', StandardScaler(), num_cols),
    ('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False), cat_cols)
], remainder='drop')

# Fit preprocessor on training data
X_train = train[num_cols + cat_cols].copy()
preprocessor.fit(X_train)
joblib.dump(preprocessor, os.path.join(MODELS_DIR, 'preprocessor.pkl'))
print('Preprocessor saved to models/preprocessor.pkl')
```

## 5) Model Training
## 📘 MODEL TRAINING — Linear | Decision Tree | Random Forest | XGBoost
Train Linear Regression, Decision Tree, RandomForest and XGBoost (per-target or multi-output). Save models.

```python
# ============================================================
# 🧠 TRAIN ALL MODELS (Including Tuned XGBoost)
# ============================================================
from sklearn.multioutput import MultiOutputRegressor
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor
from sklearn.model_selection import GridSearchCV
import joblib, xgboost as xgb
import numpy as np, os

# ✅ Ensure directories exist
os.makedirs(MODELS_DIR, exist_ok=True)

# --- Transform input features ---
X_train = preprocessor.transform(train[num_cols + cat_cols])
X_test = preprocessor.transform(test[num_cols + cat_cols])

# --- Prepare target arrays ---
if len(TARGETS) == 0:
    raise ValueError('❌ No target columns found. Please update TARGETS list.')

Y_train = train[TARGETS].values
Y_test = test[TARGETS].values


# ============================================================
# 1️⃣ LINEAR REGRESSION (Multi-Output)
# ============================================================
print("🔹 Training Linear Regression...")
lr = MultiOutputRegressor(LinearRegression())
lr.fit(X_train, Y_train)
joblib.dump(lr, os.path.join(MODELS_DIR, 'linear_multioutput.pkl'))
print("✅ Linear Regression model saved.\n")


# ============================================================
# 2️⃣ DECISION TREE REGRESSOR
# ============================================================
print("🔹 Training Decision Tree...")
dt = MultiOutputRegressor(DecisionTreeRegressor(random_state=42))
dt.fit(X_train, Y_train)
joblib.dump(dt, os.path.join(MODELS_DIR, 'dt_multioutput.pkl'))
print("✅ Decision Tree model saved.\n")


# ============================================================
# 3️⃣ RANDOM FOREST REGRESSOR
# ============================================================
print("🔹 Training Random Forest...")
rf = MultiOutputRegressor(RandomForestRegressor(
    n_estimators=200, random_state=42, n_jobs=-1))
rf.fit(X_train, Y_train)
joblib.dump(rf, os.path.join(MODELS_DIR, 'rf_multioutput.pkl'))
print("✅ Random Forest model saved.\n")


# ============================================================
# 4️⃣ XGBOOST (One Model per Target)
# ============================================================
print("🔹 Training XGBoost Models (per target)...")
xgb_models = {}

for i, target in enumerate(TARGETS):
    print(f"   ➜ Training XGB for target: {target}")
    model = XGBRegressor(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=6,
        random_state=42,
        eval_metric='rmse'
    )
    try:
        model.fit(
            X_train, Y_train[:, i],
            eval_set=[(X_test, Y_test[:, i])],
            early_stopping_rounds=25,
            verbose=False
        )
        joblib.dump(model, os.path.join(MODELS_DIR, f'xgb_{target}.pkl'))
        xgb_models[target] = model
    except TypeError:
        dtrain = xgb.DMatrix(X_train, label=Y_train[:, i])
        dtest = xgb.DMatrix(X_test, label=Y_test[:, i])
        params = {
            "objective": "reg:squarederror",
            "eval_metric": "rmse",
            "eta": 0.05,
            "max_depth": 6,
            "seed": 42
        }
        booster = xgb.train(
            params, dtrain,
            num_boost_round=300,
            evals=[(dtest, 'eval')],
            early_stopping_rounds=25
        )
        booster.save_model(os.path.join(MODELS_DIR, f'xgb_{target}.json'))
        xgb_models[target] = booster

    print(f"      ✅ Done training {target}")

# ✅ Bundle all XGBoost models
bundle_path = os.path.join(MODELS_DIR, 'xgb_models_bundle.pkl')
joblib.dump(xgb_models, bundle_path)
print(f"\n✅ All XGBoost models saved & bundled at: {bundle_path}")


# ============================================================
# 5️⃣ XGBOOST (TUNED — MultiOutput + GridSearchCV)
# ============================================================
print("\n🔹 Training Tuned XGBoost Model (global optimization)...")

from sklearn.model_selection import GridSearchCV

param_grid = {
    "n_estimators": [200, 300],
    "learning_rate": [0.05, 0.1],
    "max_depth": [4, 6, 8],
    "subsample": [0.8, 1.0],
    "colsample_bytree": [0.8, 1.0],
    "reg_lambda": [1, 1.5]
}

xgb_base = XGBRegressor(
    objective='reg:squarederror',
    eval_metric='rmse',
    random_state=42,
    n_jobs=-1
)

xgb_tuned = MultiOutputRegressor(
    GridSearchCV(xgb_base, param_grid, cv=3, scoring='r2', n_jobs=-1, verbose=0)
)

xgb_tuned.fit(X_train, Y_train)
joblib.dump(xgb_tuned, os.path.join(MODELS_DIR, 'xgb_tuned.pkl'))
print("✅ Tuned XGBoost model saved as: models/xgb_tuned.pkl\n")


# ============================================================
# ✅ COMPLETION
# ============================================================
print("🎉 All models (Linear, Decision Tree, Random Forest, XGBoost, XGBoost Tuned) trained and saved successfully!")
```

```python
# # ============================================================
# # 📊 MODEL COMPARISON DASHBOARD — with Tuned XGBoost
# # ============================================================

# import os, joblib, numpy as np, pandas as pd
# import matplotlib.pyplot as plt
# import seaborn as sns
# import xgboost as xgb
# from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
# from sklearn.model_selection import GridSearchCV
# from sklearn.multioutput import MultiOutputRegressor
# from xgboost import XGBRegressor

# sns.set(style="whitegrid", font_scale=1.2)

# # --- Container for results ---
# results = []

# # --- Helper function for consistent evaluation ---
# def evaluate_model(name, model, X_test, Y_test):
#     """Compute RMSE, MAE, R² and append to results list."""
#     Y_pred = model.predict(X_test)
#     rmse = np.sqrt(mean_squared_error(Y_test, Y_pred))
#     mae = mean_absolute_error(Y_test, Y_pred)
#     r2 = r2_score(Y_test, Y_pred)
#     results.append({'Model': name, 'RMSE': rmse, 'MAE': mae, 'R²': r2})
#     print(f"{name:<18} → RMSE: {rmse:8.2f} | MAE: {mae:8.2f} | R²: {r2:8.4f}")


# # ============================================================
# # 🔹 Evaluate Classical Models
# # ============================================================
# print("🔹 Evaluating Classical Models...")
# evaluate_model('Linear Regression', lr, X_test, Y_test)
# evaluate_model('Decision Tree', dt, X_test, Y_test)
# evaluate_model('Random Forest', rf, X_test, Y_test)


# # ============================================================
# # 🔹 Evaluate XGBoost Bundle (existing per-target models)
# # ============================================================
# xgb_bundle_path = os.path.join(MODELS_DIR, 'xgb_models_bundle.pkl')
# if os.path.exists(xgb_bundle_path):
#     print("\n🔹 Evaluating XGBoost Bundle...")
#     xgb_bundle = joblib.load(xgb_bundle_path)
#     xgb_preds = []

#     for t in TARGETS:
#         model = xgb_bundle.get(t)
#         if model is None:
#             continue
#         try:
#             y_pred = model.predict(X_test)
#         except:
#             y_pred = model.predict(xgb.DMatrix(X_test))
#         xgb_preds.append(y_pred)

#     if xgb_preds:
#         Y_pred_xgb = np.vstack(xgb_preds).T
#         usable_targets = min(Y_pred_xgb.shape[1], Y_test.shape[1])
#         rmse = np.sqrt(mean_squared_error(Y_test[:, :usable_targets], Y_pred_xgb[:, :usable_targets]))
#         mae = mean_absolute_error(Y_test[:, :usable_targets], Y_pred_xgb[:, :usable_targets])
#         r2 = r2_score(Y_test[:, :usable_targets], Y_pred_xgb[:, :usable_targets])
#         results.append({'Model': 'XGBoost', 'RMSE': rmse, 'MAE': mae, 'R²': r2})
#         print(f"{'XGBoost':<18} → RMSE: {rmse:8.2f} | MAE: {mae:8.2f} | R²: {r2:8.4f}")
# else:
#     print("\n⚠️ No XGBoost bundle found at:", xgb_bundle_path)


# # ============================================================
# # 🔹 Train & Evaluate XGBoost (Tuned)
# # ============================================================
# xgb_tuned_path = os.path.join(MODELS_DIR, 'xgb_tuned.pkl')

# if not os.path.exists(xgb_tuned_path):
#     print("\n🔹 Training XGBoost (Tuned) Model...")
#     param_grid = {
#         "n_estimators": [200, 300],
#         "learning_rate": [0.05, 0.1],
#         "max_depth": [4, 6, 8],
#         "subsample": [0.8, 1.0],
#         "colsample_bytree": [0.8, 1.0],
#         "reg_lambda": [1, 1.5]
#     }

#     xgb_base = XGBRegressor(
#         objective='reg:squarederror',
#         eval_metric='rmse',
#         random_state=42,
#         n_jobs=-1
#     )

#     xgb_tuned = MultiOutputRegressor(
#         GridSearchCV(xgb_base, param_grid, cv=3, scoring='r2', n_jobs=-1, verbose=0)
#     )

#     xgb_tuned.fit(X_train, Y_train)
#     joblib.dump(xgb_tuned, xgb_tuned_path)
#     print("✅ Tuned XGBoost model saved as:", xgb_tuned_path)
# else:
#     print("\n✅ Tuned XGBoost model found. Skipping retraining.")

# # Evaluate Tuned XGBoost
# xgb_tuned = joblib.load(xgb_tuned_path)
# evaluate_model('XGBoost (Tuned)', xgb_tuned, X_test, Y_test)


# # ============================================================
# # 📋 Results Summary
# # ============================================================
# results_df = pd.DataFrame(results).sort_values(by='R²', ascending=False).reset_index(drop=True)

# print("\n📊 Model Performance Summary:")
# display(
#     results_df.style
#     .background_gradient(cmap='Greens', subset=['R²'])
#     .background_gradient(cmap='Oranges', subset=['RMSE'])
#     .background_gradient(cmap='Blues', subset=['MAE'])
#     .format({'RMSE': '{:.2f}', 'MAE': '{:.2f}', 'R²': '{:.4f}'})
# )


# # ============================================================
# # 🎨 Visualization Dashboard
# # ============================================================
# fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# # --- RMSE Chart ---
# sns.barplot(y='Model', x='RMSE', hue='Model', dodge=False, data=results_df, ax=axes[0], legend=False, palette='Blues_r')
# axes[0].set_title('RMSE ↓ (Lower is Better)', fontsize=12, weight='bold')
# axes[0].bar_label(axes[0].containers[0], fmt='%.2f', label_type='edge')

# # --- MAE Chart ---
# sns.barplot(y='Model', x='MAE', hue='Model', dodge=False, data=results_df, ax=axes[1], legend=False, palette='Oranges_r')
# axes[1].set_title('MAE ↓ (Lower is Better)', fontsize=12, weight='bold')
# axes[1].bar_label(axes[1].containers[0], fmt='%.2f', label_type='edge')

# # --- R² Chart ---
# sns.barplot(y='Model', x='R²', hue='Model', dodge=False, data=results_df, ax=axes[2], legend=False, palette='Greens')
# axes[2].set_title('R² ↑ (Higher is Better)', fontsize=12, weight='bold')
# axes[2].bar_label(axes[2].containers[0], fmt='%.3f', label_type='edge')

# plt.suptitle('📈 Model Performance Comparison — Linear | Tree | Forest | XGBoost | Tuned XGBoost', 
#              fontsize=16, weight='bold')
# plt.tight_layout(rect=[0, 0, 1, 0.95])
# plt.show()


# # ============================================================
# # 🏆 Identify and Save Best Model
# # ============================================================
# best = results_df.loc[results_df['R²'].idxmax()]
# print(f"\n🏆 Best Overall Model: {best['Model']} | R²={best['R²']:.4f} | RMSE={best['RMSE']:.2f} | MAE={best['MAE']:.2f}")

# model_file_map = {
#     'Linear Regression': 'linear_multioutput.pkl',
#     'Decision Tree': 'dt_multioutput.pkl',
#     'Random Forest': 'rf_multioutput.pkl',
#     'XGBoost': 'xgb_models_bundle.pkl',
#     'XGBoost (Tuned)': 'xgb_tuned.pkl'
# }

# best_model_file = model_file_map.get(best['Model'])
# if best_model_file:
#     src_path = os.path.join(MODELS_DIR, best_model_file)
#     dst_path = os.path.join(MODELS_DIR, 'best_model.pkl')
#     if os.path.exists(src_path):
#         joblib.dump(joblib.load(src_path), dst_path)
#         print(f"✅ Best model ({best['Model']}) saved as: {dst_path}")
#     else:
#         print(f"⚠️ Model file not found for {best['Model']}, skipping save.")
# else:
#     print("⚠️ No matching model found for saving best model.")

# ============================================================
# 📊 MODEL COMPARISON DASHBOARD — Enhanced (with Tuned XGBoost)
# ============================================================

import os, joblib, numpy as np, pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import xgboost as xgb
from tqdm import tqdm
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import GridSearchCV
from sklearn.multioutput import MultiOutputRegressor
from xgboost import XGBRegressor

sns.set(style="whitegrid", font_scale=1.2)

# --- Container for results ---
results = []

# --- Helper function for consistent evaluation ---
def evaluate_model(name, model, X_test, Y_test):
    """Compute RMSE, MAE, R² and append to results list."""
    try:
        Y_pred = model.predict(X_test)
        rmse = np.sqrt(mean_squared_error(Y_test, Y_pred))
        mae = mean_absolute_error(Y_test, Y_pred)
        r2 = r2_score(Y_test, Y_pred)
        results.append({'Model': name, 'RMSE': rmse, 'MAE': mae, 'R²': r2})
        print(f"{name:<20} → RMSE: {rmse:8.2f} | MAE: {mae:8.2f} | R²: {r2:8.4f}")
    except Exception as e:
        print(f"⚠️ Error evaluating {name}: {e}")


# ============================================================
# 🔹 Evaluate Classical Models
# ============================================================
print("🔹 Evaluating Classical Models...")
evaluate_model('Linear Regression', lr, X_test, Y_test)
evaluate_model('Decision Tree', dt, X_test, Y_test)
evaluate_model('Random Forest', rf, X_test, Y_test)


# ============================================================
# 🔹 Evaluate XGBoost Bundle (existing per-target models)
# ============================================================
xgb_bundle_path = os.path.join(MODELS_DIR, 'xgb_models_bundle.pkl')
if os.path.exists(xgb_bundle_path):
    print("\n🔹 Evaluating XGBoost Bundle...")
    xgb_bundle = joblib.load(xgb_bundle_path)
    xgb_preds = []

    for t in tqdm(TARGETS, desc="Evaluating XGBoost Targets"):
        model = xgb_bundle.get(t)
        if model is None:
            continue
        try:
            y_pred = model.predict(X_test)
        except:
            y_pred = model.predict(xgb.DMatrix(X_test))
        xgb_preds.append(y_pred)

    if xgb_preds:
        Y_pred_xgb = np.vstack(xgb_preds).T
        usable_targets = min(Y_pred_xgb.shape[1], Y_test.shape[1])
        rmse = np.sqrt(mean_squared_error(Y_test[:, :usable_targets], Y_pred_xgb[:, :usable_targets]))
        mae = mean_absolute_error(Y_test[:, :usable_targets], Y_pred_xgb[:, :usable_targets])
        r2 = r2_score(Y_test[:, :usable_targets], Y_pred_xgb[:, :usable_targets])
        results.append({'Model': 'XGBoost', 'RMSE': rmse, 'MAE': mae, 'R²': r2})
        print(f"{'XGBoost':<20} → RMSE: {rmse:8.2f} | MAE: {mae:8.2f} | R²: {r2:8.4f}")
else:
    print("\n⚠️ No XGBoost bundle found at:", xgb_bundle_path)


# ============================================================
# 🔹 Train & Evaluate XGBoost (Tuned)
# ============================================================
xgb_tuned_path = os.path.join(MODELS_DIR, 'xgb_tuned.pkl')

if not os.path.exists(xgb_tuned_path):
    print("\n🔹 Training XGBoost (Tuned) Model with GridSearchCV...")

    param_grid = {
        "n_estimators": [200, 300],
        "learning_rate": [0.05, 0.1],
        "max_depth": [4, 6, 8],
        "subsample": [0.8, 1.0],
        "colsample_bytree": [0.8, 1.0],
        "reg_lambda": [1, 1.5]
    }

    xgb_base = XGBRegressor(
        objective='reg:squarederror',
        eval_metric='rmse',
        random_state=42,
        n_jobs=-1
    )

    xgb_tuned = MultiOutputRegressor(
        GridSearchCV(xgb_base, param_grid, cv=3, scoring='r2', n_jobs=-1, verbose=1)
    )

    xgb_tuned.fit(X_train, Y_train)
    joblib.dump(xgb_tuned, xgb_tuned_path)
    print(f"✅ Tuned XGBoost model saved as: {xgb_tuned_path}")
else:
    print("\n✅ Tuned XGBoost model found. Skipping retraining.")

# Evaluate Tuned XGBoost
try:
    xgb_tuned = joblib.load(xgb_tuned_path)
    evaluate_model('XGBoost (Tuned)', xgb_tuned, X_test, Y_test)
except Exception as e:
    print(f"⚠️ Unable to evaluate Tuned XGBoost: {e}")


# ============================================================
# 📋 Results Summary
# ============================================================
results_df = pd.DataFrame(results).sort_values(by='R²', ascending=False).reset_index(drop=True)

print("\n📊 Model Performance Summary:")
print(results_df.round(4))
results_df.to_csv(os.path.join(MODELS_DIR, "model_metrics.csv"), index=False)
print(f"📁 Saved metrics report to: {os.path.join(MODELS_DIR, 'model_metrics.csv')}")

# ============================================================
# 🎨 Visualization Dashboard
# ============================================================
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

colors_rmse = sns.color_palette("Blues_r", len(results_df))
colors_mae = sns.color_palette("Oranges_r", len(results_df))
colors_r2 = sns.color_palette("Greens", len(results_df))

sns.barplot(y='Model', x='RMSE', data=results_df, ax=axes[0], palette=colors_rmse)
axes[0].set_title('RMSE ↓ (Lower is Better)', fontsize=13, weight='bold')
axes[0].bar_label(axes[0].containers[0], fmt='%.2f', label_type='edge')

sns.barplot(y='Model', x='MAE', data=results_df, ax=axes[1], palette=colors_mae)
axes[1].set_title('MAE ↓ (Lower is Better)', fontsize=13, weight='bold')
axes[1].bar_label(axes[1].containers[0], fmt='%.2f', label_type='edge')

sns.barplot(y='Model', x='R²', data=results_df, ax=axes[2], palette=colors_r2)
axes[2].set_title('R² ↑ (Higher is Better)', fontsize=13, weight='bold')
axes[2].bar_label(axes[2].containers[0], fmt='%.3f', label_type='edge')

plt.suptitle('📈 Model Performance Comparison — Linear | Tree | Forest | XGBoost | Tuned XGBoost',
             fontsize=16, weight='bold')
plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()

# ============================================================
# 🏆 Identify and Save Best Model
# ============================================================
best = results_df.loc[results_df['R²'].idxmax()]
print(f"\n🏆 Best Overall Model: {best['Model']} | R²={best['R²']:.4f} | RMSE={best['RMSE']:.2f} | MAE={best['MAE']:.2f}")

model_file_map = {
    'Linear Regression': 'linear_multioutput.pkl',
    'Decision Tree': 'dt_multioutput.pkl',
    'Random Forest': 'rf_multioutput.pkl',
    'XGBoost': 'xgb_models_bundle.pkl',
    'XGBoost (Tuned)': 'xgb_tuned.pkl'
}

best_model_file = model_file_map.get(best['Model'])
if best_model_file:
    src_path = os.path.join(MODELS_DIR, best_model_file)
    dst_path = os.path.join(MODELS_DIR, 'best_model.pkl')
    if os.path.exists(src_path):
        joblib.dump(joblib.load(src_path), dst_path)
        print(f"✅ Best model ({best['Model']}) saved as: {dst_path}")
    else:
        print(f"⚠️ Model file not found for {best['Model']}, skipping save.")
else:
    print("⚠️ No matching model found for saving best model.")
```

```python
# # ============================================================
# # 🧠 Predict Budget from Selected Model (Final Production Version)
# # ============================================================

# import os, joblib, numpy as np, pandas as pd
# import matplotlib.pyplot as plt
# import xgboost as xgb

# def predict_budget_from_model(user_dict, model_name='RandomForest'):
#     """
#     Predicts monthly budget allocation for a user using a trained model.
#     Supports: Linear, DecisionTree, RandomForest, and XGBoost.
#     """

#     # --- Load the requested model ---
#     model_path_map = {
#         'Linear': os.path.join(MODELS_DIR, 'linear_multioutput.pkl'),
#         'DecisionTree': os.path.join(MODELS_DIR, 'dt_multioutput.pkl'),
#         'RandomForest': os.path.join(MODELS_DIR, 'rf_multioutput.pkl'),
#         'XGBoost': os.path.join(MODELS_DIR, 'xgb_models_bundle.pkl')
#     }

#     if model_name not in model_path_map:
#         raise ValueError(f"❌ Unknown model '{model_name}'. Choose from {list(model_path_map.keys())}")

#     model_path = model_path_map[model_name]
#     if not os.path.exists(model_path):
#         raise FileNotFoundError(f"⚠️ Model file not found: {model_path}")

#     model = joblib.load(model_path)
#     print(f"\n✅ Using {model_name} model")

#     # --- Convert user input to DataFrame ---
#     df_user = pd.DataFrame([user_dict])

#     # --- Ensure numeric columns ---
#     numeric_cols = ['Age', 'Income', 'Dependents', 'debt', 
#                     'debt_to_income_ratio', 'expense_ratio', 'Loan_Repayment']
#     for col in numeric_cols:
#         if col not in df_user.columns:
#             df_user[col] = 0.0

#     # --- Derived features ---
#     df_user['debt_to_income_ratio'] = (
#         df_user['debt'] / df_user['Income'].replace(0, np.nan)
#     ).replace([np.inf, -np.inf], 0).fillna(0)

#     # Default expense ratio if missing
#     df_user['expense_ratio'] = df_user.get('expense_ratio', 0.0)

#     # --- Compute EMI (Loan Repayment) if missing ---
#     def compute_emi(principal, annual_rate=0.12, years=3):
#         try:
#             p = float(principal)
#         except:
#             return 0.0
#         if p <= 0:
#             return 0.0
#         r = annual_rate / 12
#         n = years * 12
#         return round((p * r * (1 + r)**n) / ((1 + r)**n - 1), 2)

#     if 'Loan_Repayment' not in df_user.columns or df_user['Loan_Repayment'].iloc[0] == 0:
#         df_user['Loan_Repayment'] = df_user['debt'].apply(compute_emi)

#     # --- Ensure categorical columns ---
#     cat_cols = ['Occupation', 'City_Tier']
#     for col in cat_cols:
#         if col not in df_user.columns:
#             df_user[col] = "Unknown"

#     # --- Align with preprocessor features ---
#     preprocessor_features = preprocessor.feature_names_in_
#     for col in preprocessor_features:
#         if col not in df_user.columns:
#             df_user[col] = 0.0

#     # --- Transform numeric/categorical data ---
#     Xu = preprocessor.transform(df_user[preprocessor_features])

#     # --- Make Predictions ---
#     if model_name == 'XGBoost':
#         preds = []
#         for t in TARGETS:
#             if t in model:
#                 m = model[t]
#                 try:
#                     y = m.predict(Xu)
#                 except:
#                     y = m.predict(xgb.DMatrix(Xu))
#                 preds.append(y)
#         ypred = np.array(preds).flatten()
#     else:
#         ypred = model.predict(Xu)[0]

#     # --- Map to target categories ---
#     pred = dict(zip(TARGETS, [max(0, round(float(v), 2)) for v in ypred]))

#     # --- Compute Savings logically ---
#     others = sum(v for k, v in pred.items() if k.lower() != 'savings')
#     pred['Savings'] = max(0, round(user_dict['Income'] - others, 2))

#     # --- Print textual summary ---
#     print("\n📊 Budget Breakdown:")
#     for k, v in pred.items():
#         print(f"  {k:<15}: ₹{v:,.2f}")

#     # --- Visualization: Horizontal Bar Chart ---
#     plt.figure(figsize=(8, 5))
#     bars = plt.barh(list(pred.keys()), list(pred.values()), color='teal', edgecolor='black', alpha=0.8)
#     plt.xlabel("Amount (₹)", fontsize=12)
#     plt.title(f"Predicted Budget Breakdown — {model_name}", fontsize=14, weight='bold')

#     # Add labels beside bars
#     for bar in bars:
#         plt.text(bar.get_width() + user_dict['Income'] * 0.01, bar.get_y() + bar.get_height()/2,
#                  f"₹{bar.get_width():,.0f}", va='center', fontsize=10)

#     plt.tight_layout()
#     plt.show()

#     # --- Display savings ratio ---
#     savings_ratio = (pred['Savings'] / user_dict['Income']) * 100
#     print(f"\n💰 Estimated Savings Ratio: {savings_ratio:.2f}% of income")

#     return pred


# # ============================================================
# # ✅ TEST CASES
# # ============================================================

# user_with_loan = {
#     'Age': 28, 'Income': 60000, 'Dependents': 2,
#     'Occupation': 'IT Professional', 'City_Tier': 'Tier_1', 'debt': 50000
# }

# user_no_loan = {
#     'Age': 21, 'Income': 40000, 'Dependents': 1,
#     'Occupation': 'IT Professional', 'City_Tier': 'Tier_3', 'debt': 0
# }

# print("\n📊 Prediction: With Loan (RandomForest)")
# predict_budget_from_model(user_with_loan, model_name='RandomForest')

# print("\n📊 Prediction: No Loan (XGBoost)")
# predict_budget_from_model(user_no_loan, model_name='XGBoost')


# ============================================================
# 🤖 Smart Budget Predictor (Auto Model Selection)
# ============================================================

import os, joblib, numpy as np, pandas as pd
import matplotlib.pyplot as plt
import xgboost as xgb

def predict_budget_from_model(user_dict):
    """
    Predicts monthly budget allocation for a user using an *auto-selected* model.
    💡 RandomForest → if user has debt
    💡 XGBoost → if user has no debt
    """

    # --- Auto model selection based on debt presence ---
    model_name = 'RandomForest' if user_dict.get('debt', 0) > 0 else 'XGBoost'
    print(f"\n🧠 Auto-selected model based on user's financial profile: {model_name}")

    # --- Define model paths ---
    model_path_map = {
        'Linear': os.path.join(MODELS_DIR, 'linear_multioutput.pkl'),
        'DecisionTree': os.path.join(MODELS_DIR, 'dt_multioutput.pkl'),
        'RandomForest': os.path.join(MODELS_DIR, 'rf_multioutput.pkl'),
        'XGBoost': os.path.join(MODELS_DIR, 'xgb_models_bundle.pkl')
    }

    model_path = model_path_map[model_name]
    if not os.path.exists(model_path):
        raise FileNotFoundError(f"⚠️ Model file not found: {model_path}")

    model = joblib.load(model_path)
    print(f"✅ Using {model_name} model")

    # --- Convert user input to DataFrame ---
    df_user = pd.DataFrame([user_dict])

    # --- Ensure numeric columns ---
    numeric_cols = ['Age', 'Income', 'Dependents', 'debt', 
                    'debt_to_income_ratio', 'expense_ratio', 'Loan_Repayment']
    for col in numeric_cols:
        if col not in df_user.columns:
            df_user[col] = 0.0

    # --- Derived features ---
    df_user['debt_to_income_ratio'] = (
        df_user['debt'] / df_user['Income'].replace(0, np.nan)
    ).replace([np.inf, -np.inf], 0).fillna(0)

    # Default expense ratio if missing
    df_user['expense_ratio'] = df_user.get('expense_ratio', 0.0)

    # --- Compute EMI (Loan Repayment) if missing ---
    def compute_emi(principal, annual_rate=0.12, years=3):
        try:
            p = float(principal)
        except:
            return 0.0
        if p <= 0:
            return 0.0
        r = annual_rate / 12
        n = years * 12
        return round((p * r * (1 + r)**n) / ((1 + r)**n - 1), 2)

    if 'Loan_Repayment' not in df_user.columns or df_user['Loan_Repayment'].iloc[0] == 0:
        df_user['Loan_Repayment'] = df_user['debt'].apply(compute_emi)

    # --- Ensure categorical columns ---
    cat_cols = ['Occupation', 'City_Tier']
    for col in cat_cols:
        if col not in df_user.columns:
            df_user[col] = "Unknown"

    # --- Align with preprocessor features ---
    preprocessor_features = preprocessor.feature_names_in_
    for col in preprocessor_features:
        if col not in df_user.columns:
            df_user[col] = 0.0

    # --- Transform numeric/categorical data ---
    Xu = preprocessor.transform(df_user[preprocessor_features])

    # --- Make Predictions ---
    if model_name == 'XGBoost':
        preds = []
        for t in TARGETS:
            if t in model:
                m = model[t]
                try:
                    y = m.predict(Xu)
                except:
                    y = m.predict(xgb.DMatrix(Xu))
                preds.append(y)
        ypred = np.array(preds).flatten()
    else:
        ypred = model.predict(Xu)[0]

    # --- Map to target categories ---
    pred = dict(zip(TARGETS, [max(0, round(float(v), 2)) for v in ypred]))

    # --- Compute Savings logically ---
    others = sum(v for k, v in pred.items() if k.lower() != 'savings')
    pred['Savings'] = max(0, round(user_dict['Income'] - others, 2))

    # --- Print textual summary ---
    print("\n📊 Budget Breakdown:")
    for k, v in pred.items():
        print(f"  {k:<15}: ₹{v:,.2f}")

    # --- Visualization: Horizontal Bar Chart ---
    plt.figure(figsize=(8, 5))
    bars = plt.barh(list(pred.keys()), list(pred.values()), color='teal', edgecolor='black', alpha=0.8)
    plt.xlabel("Amount (₹)", fontsize=12)
    plt.title(f"Predicted Budget Breakdown — {model_name}", fontsize=14, weight='bold')

    # Add labels beside bars
    for bar in bars:
        plt.text(bar.get_width() + user_dict['Income'] * 0.01, bar.get_y() + bar.get_height()/2,
                 f"₹{bar.get_width():,.0f}", va='center', fontsize=10)

    plt.tight_layout()
    plt.show()
    # --- Display savings ratio ---
    savings_ratio = (pred['Savings'] / user_dict['Income']) * 100
    print(f"\n💰 Estimated Savings Ratio: {savings_ratio:.2f}% of income")

    return pred

# ============================================================
# ✅ TEST CASES
# ============================================================

user_with_loan = {
    'Age': 28, 'Income': 60000, 'Dependents': 1,
    'Occupation': 'IT Professional', 'City_Tier': 'Tier_2', 'debt': 50000
}

user_no_loan = {
    'Age': 21, 'Income': 40000, 'Dependents': 1,
    'Occupation': 'IT Professional', 'City_Tier': 'Tier_3', 'debt': 0
}

print("\n📊 Prediction: With Loan (Auto Select)")
predict_budget_from_model(user_with_loan)

print("\n📊 Prediction: No Loan (Auto Select)")
predict_budget_from_model(user_no_loan)

```
